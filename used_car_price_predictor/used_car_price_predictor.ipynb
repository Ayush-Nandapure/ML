```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns = ["torque"])
```

```python
df.head()
```

```python
df.shape
```

```python
df['brand'] = df['name'].str.extract(r'^(\w+)')
```

```python
df['mileage'] = df['mileage'].str.extract(r'(\d+\.?\d*)').astype(float)
```

```python
df.head()
```

```python
df['engine'] = df['engine'].str.extract(r'(\d+\.?\d*)').astype(float)
df['max_power'] = df['max_power'].str.extract(r'(\d+\.?\d*)').astype(float)
```

```python
df = df.drop(columns = ['name'])
```

```python
df = df[df['seller_type'] != 'Trustmark Dealer']
df = df[df['owner'] != 'Test Drive Car']
df = df[(df['fuel'] != "LPG") & (df['fuel'] != "CNG")]
```

```python
threshold = 30
counts = df['brand'].value_counts()
repl = counts[counts < threshold].index
df['brand'] = df['brand'].replace(repl, 'others')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('selling_price', axis=1), df['selling_price'], test_size=0.2, random_state=42)
```

```python
_75_percentile = df['selling_price'].quantile(0.75)
_25_percentile = df['selling_price'].quantile(0.25)
min = _25_percentile - 1.5 * (_75_percentile - _25_percentile)
max = _75_percentile + 1.5 * (_75_percentile - _25_percentile)
```

```python
y_train = y_train[(y_train >= min) & (y_train <= max)]
X_train = X_train.loc[y_train.index]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import PowerTransformer
from sklearn.impute import SimpleImputer
```

```python
pipe1 = Pipeline([
    ('imputer', KNNImputer(n_neighbors = 3, weights = 'distance')),
    ('scaler', StandardScaler()),
    ('normalise', PowerTransformer())
])
```

```python
pipe2 = Pipeline([
    ('scaler2', StandardScaler()),
    ("normalise2", PowerTransformer())
])
```

```python
pipe3 = Pipeline([
    ('ohe', OneHotEncoder(sparse_output = False, drop = 'first'))
])
```

```python
pipe4 = Pipeline([
    ('imputer2', SimpleImputer(strategy = 'most_frequent')),
    ('ohe2', OneHotEncoder(sparse_output = False, drop = 'first', min_frequency=0.05,))
])
```

```python
trf = ColumnTransformer([
    ('1', pipe1, ['mileage', 'engine', 'max_power']),
    ('2', pipe2, ['year', 'km_driven']),
    ('3', pipe3, ['fuel', 'seller_type', 'transmission', 'owner', 'brand']),
    ('4', pipe4, ['seats'])
], remainder = 'passthrough')
```

```python
main_pipe = Pipeline([
    ('trf1', trf),
    ('lr', LinearRegression())
])
```

```python
main_pipe.fit(X_train,y_train)
```

```python
pred = main_pipe.predict(X_test)
```

```python
mse = mean_squared_error(y_test, pred)
r2 = r2_score(y_test, pred)
```
